```python
import random

seed_val = 1
random.seed(seed_val)

raw_readings = [round(random.uniform(-50.0, 150.0), 2) for _ in range(5)]
sensor_data = raw_readings + ["INVALID_SENSOR"]

valid_readings = []
invalid_count = 0
classification_results = {"Normal": 0, "Warning": 0, "Critical": 0}

print(f"Generated Sensor Data: {sensor_data}")

for reading in sensor_data:
    try:
        val = float(reading)
        if 0.0 <= val <= 50.0:
            classification_results["Normal"] += 1
        elif 50.1 <= val <= 100.0:
            classification_results["Warning"] += 1
        else:
            classification_results["Critical"] += 1
        valid_readings.append(val)
    except ValueError:
        invalid_count += 1

print(f"Valid/Invalid Results: Valid: {len(valid_readings)}, Invalid: {invalid_count}")
print(f"Classification Results: {classification_results}")
print("Execution Log: Processed readings successfully with exception handling applied.")
print("Final Output: Sensor summary generated based on SEED_NUM = 1 with active range validation.")
```
