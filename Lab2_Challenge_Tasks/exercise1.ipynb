```python
import random

# --- STUDENT ASSESSMENT HEADER ---
surname, student_id, seed = "dacua", "TUPM-26-3495", 1
random.seed(seed + len(surname))
print(f"Student: {surname.upper()} | ID: {student_id} | Seed: {seed}")
print("=" * 50)

# Assessment Data
raw = [str(round(random.uniform(10, 100), 2)), "ERR", "55.5", "INVALID"]
valid, class_res, log = [], [], []

for i, item in enumerate(raw):
  try:
    v = float(item)
    valid.append((i, v))
    c = "LOW" if v < 40 else ("NORMAL" if v <= 80 else "HIGH")
    class_res.append(f"Reading {i}({v}):{c}")
    log.append(f"Entry {i}:Valid")
  except ValueError:
    log.append(f"Entry {i}:Invalid")

final_output = f"Processed {len(valid)} valid sensor readings out of {len(raw)}."

# Print Assessment Data & Final Output
print("=== ASSESSMENT DATA: EXERCISE 1 ===")
print("Generated Sensor Data:", raw)
print(f"Valid/Invalid Results: {len(valid)} valid, {len(raw)-len(valid)} invalid")
print("Classification Results:", class_res)
print("Execution Log:", log)
print("Final Output:", final_output)
```
