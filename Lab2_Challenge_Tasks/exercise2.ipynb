```python
import random

# Paggamit ng seed = 1 para sa Exercise 2
seed_val = 1
random.seed(seed_val)

# 1. Generate signal base sa apelyido at seed
chars = "TUPM-26-3495#SEC"
generated_signal = "".join(random.choices(chars, k=12))

# 2. Process and normalize signal (alisin ang mga spaces kung meron)
processed_signal = generated_signal.replace(" ", "")

# 3. Analyze character by character
letters = 0
numbers = 0
symbols = 0

for char in processed_signal:
    if char.isalpha():
        letters += 1
    elif char.isdigit():
        numbers += 1
    else:
        symbols += 1

# 4. Display analysis and classification
print(f"Generated Signal: {generated_signal}")
print(f"Processed Signal: {processed_signal}")
print(f"Character Analysis: Letters: {letters}, Numbers: {numbers}, Symbols: {symbols}")
print("Signal Classification: Alphanumeric Mixed Signal")
print("Execution Log: Signal split and analyzed successfully per character.")
print("Final Output: Diagnostic report compiled successfully.")
```
