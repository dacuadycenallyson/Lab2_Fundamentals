```python
import random
import string

# --- STUDENT ASSESSMENT HEADER ---
surname, student_id, seed = "dacua", "TUPM-26-3495", 1
random.seed(seed + len(surname))
print(f"Student: {surname.upper()} | ID: {student_id} | Seed: {seed}")
print("=" * 50)

# Assessment Data
gen_signal = "".join(random.choices(surname + string.ascii_letters, k=6))
proc_signal = gen_signal.upper()
char_analysis = {c: proc_signal.count(c) for c in set(proc_signal)}
classification = "ALPHANUMERIC"
exec_log = "Signal analyzed successfully."
final_output = f"Signal: {proc_signal} | Class: {classification}"

# Print Assessment Data & Final Output
print("=== ASSESSMENT DATA: EXERCISE 2 ===")
print("Generated Signal:", gen_signal)
print("Processed Signal:", proc_signal)
print("Character Analysis:", char_analysis)
print("Signal Classification:", classification)
print("Execution Log:", exec_log)
print("Final Output:", final_output)
```
