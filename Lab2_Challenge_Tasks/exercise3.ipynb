```python
import random

# Paggamit ng seed = 1 para sa Exercise 3
seed_val = 1
random.seed(seed_val)

# 1. Generate system performance scores (gawa ng random scores mula 60 hanggang 100)
performance_scores = [random.randint(60, 100) for _ in range(6)]

# 2. Process and compute statistics
total_scores = len(performance_scores)
average_score = sum(performance_scores) / total_scores
highest_score = max(performance_scores)
lowest_score = min(performance_scores)

# 3. Classify overall system status
if average_score >= 85:
    system_status = "Optimal Performance"
elif average_score >= 75:
    system_status = "Stable Performance"
else:
    system_status = "Needs Maintenance"

# 4. Display final system report
print(f"Generated Performance Scores: {performance_scores}")
print(f"Total Systems Analyzed: {total_scores}")
print(f"Average Score: {average_score:.2f}")
print(f"Highest Score: {highest_score}, Lowest Score: {lowest_score}")
print(f"System Status Classification: {system_status}")
print("Execution Log: Performance metrics computed and evaluated successfully.")
print("Final Output: Challenge task summary report generated.")
```
