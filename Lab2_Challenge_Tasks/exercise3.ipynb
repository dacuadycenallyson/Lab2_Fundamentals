```python
import random
import string

# --- STUDENT ASSESSMENT HEADER ---
surname, student_id, seed = "dacua", "TUPM-26-3495", 1
random.seed(seed + len(surname))
print(f"Student: {surname.upper()} | ID: {student_id} | Seed: {seed}")
print("=" * 50)

# 1. Generated Password (using surname and seed)
chars = surname + string.ascii_letters + string.digits
generated_password = "".join(random.choices(chars, k=6))

# 2. Attempt Limit (Derived dynamically from seed and surname length)
attempt_limit = 3 + ((seed + len(surname)) % 3)

# 3. Attempts Made & Access Result
attempts_made = 1  # Simulated successful attempt within the limit
access_result = "GRANTED" if attempts_made <= attempt_limit else "DENIED"

# 4. Final System State
final_system_state = "UNLOCKED" if access_result == "GRANTED" else "LOCKED"

# 5. Execution Log & Final Output
exec_log = (
    f"Attempt {attempts_made}/{attempt_limit}: Credentials verified successfully."
)
final_output = (
    f"Password: {generated_password} | State: {final_system_state}"
)

# Print Assessment Data (Exact Match to Lab Sheet)
print("=== ASSESSMENT DATA: EXERCISE 3 ===")
print("Generated Password:", generated_password)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts_made)
print("Access Result:", access_result)
print("Final System State:", final_system_state)
print("Execution Log:", exec_log)
print("Final Output:", final_output)
```
